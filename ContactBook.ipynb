## Contact Book Program

This program allows you to manage your contacts. You can add new contacts, delete existing ones, update contact details, and view all contacts in your book.

```python
# This cell is now empty as its content has been moved to the main program cell for correct execution order.
```

```python
import csv
import os # Import os module to check if file exists
import json # Import json module for JSON export

CSV_FILE = 'contacts.csv'
JSON_FILE = 'contacts.json'

class Contact:
    def __init__(self, name, phone, email):
        self.name = name
        self.phone = phone
        self.email = email

    def __str__(self):
        return f"Name: {self.name}, Phone: {self.phone}, Email: {self.email}"

    # Add a method to convert Contact object to a dictionary for JSON serialization
    def to_dict(self):
        return {
            'name': self.name,
            'phone': self.phone,
            'email': self.email
        }

def is_valid_phone(phone_number):
    """Checks if a phone number consists only of digits and has a reasonable length."""
    # Basic validation: check if it's all digits and has a length between 7 and 15 (common international range)
    return phone_number.isdigit() and 7 <= len(phone_number) <= 15

def load_contacts():
    """Loads contacts from the CSV file."""
    contacts_list = []
    if os.path.exists(CSV_FILE):
        with open(CSV_FILE, 'r', newline='') as file:
            reader = csv.reader(file)
            try:
                header = next(reader) # Skip header row
            except StopIteration:
                # File is empty, no header
                return contacts_list

            for row in reader:
                if len(row) == 3:
                    contacts_list.append(Contact(row[0], row[1], row[2]))
    return contacts_list

def save_contacts(contacts_list):
    """Saves contacts to the CSV file."""
    with open(CSV_FILE, 'w', newline='') as file:
        writer = csv.writer(file)
        writer.writerow(['Name', 'Phone', 'Email']) # Write header
        for contact in contacts_list:
            writer.writerow([contact.name, contact.phone, contact.email])

def export_contacts_to_json(contacts_list):
    """Exports contacts to a JSON file."""
    if not contacts_list:
        print("No contacts to export.")
        return

    contacts_dict_list = [contact.to_dict() for contact in contacts_list]
    with open(JSON_FILE, 'w') as json_file:
        json.dump(contacts_dict_list, json_file, indent=4)
    print(f"Contacts exported successfully to '{JSON_FILE}'.")

def add_contact(contacts_list, name, phone, email):
    """Adds a new contact to the contact list and saves to CSV."""
    new_contact = Contact(name, phone, email)
    contacts_list.append(new_contact)
    save_contacts(contacts_list) # Save contacts after adding
    print(f"Contact '{name}' added successfully.")

def delete_contact(contacts_list, name):
    """Deletes a contact by name from the contact list and saves to CSV."""
    initial_len = len(contacts_list)
    contacts_list[:] = [contact for contact in contacts_list if contact.name.lower() != name.lower()]
    if len(contacts_list) < initial_len:
        print(f"Contact '{name}' deleted successfully.")
        save_contacts(contacts_list) # Save contacts after deleting
    else:
        print(f"Contact '{name}' not found.")

def update_contact(contacts_list, name, new_phone=None, new_email=None):
    """Updates the phone and/or email for an existing contact and saves to CSV."""
    found = False
    for contact in contacts_list:
        if contact.name.lower() == name.lower():
            if new_phone:
                contact.phone = new_phone
            if new_email:
                contact.email = new_email
            print(f"Contact '{name}' updated successfully.")
            found = True
            break
    if found:
        save_contacts(contacts_list) # Save contacts after updating
    else:
        print(f"Contact '{name}' not found.")

def view_contacts(contacts_list):
    """Displays all contacts in the contact list."""
    if not contacts_list:
        print("Contact book is empty.")
    else:
        print("--- Your Contacts ---")
        for contact in contacts_list:
            print(contact)
        print("-----------------------")

# Load contacts when the program starts
contacts = load_contacts()

def main_menu():
    while True:
        print("\n--- Contact Book Menu ---")
        print("1. Add Contact")
        print("2. Delete Contact")
        print("3. Update Contact")
        print("4. View Contacts")
        print("5. Export to JSON") # New menu option
        print("6. Exit")
        choice = input("Enter your choice (1-6): ")

        if choice == '1':
            name = input("Enter contact name: ")
            phone = input("Enter contact phone number: ")
            if not is_valid_phone(phone):
                print("Invalid phone number. Please enter digits only (7-15 digits long).")
                continue
            email = input("Enter contact email address: ")
            add_contact(contacts, name, phone, email)
        elif choice == '2':
            name = input("Enter name of contact to delete: ")
            delete_contact(contacts, name)
        elif choice == '3':
            name_to_update = input("Enter name of contact to update: ")
            found_contact = None
            for contact in contacts:
                if contact.name.lower() == name_to_update.lower():
                    found_contact = contact
                    break

            if found_contact:
                print(f"\n--- Current details for '{found_contact.name}' ---")
                print(f"  Phone: {found_contact.phone}")
                print(f"  Email: {found_contact.email}")
                print("------------------------------------")

                new_phone = input(f"Enter new phone number (current: {found_contact.phone}, leave blank to keep current): ")
                if new_phone and not is_valid_phone(new_phone):
                    print("Invalid phone number. Please enter digits only (7-15 digits long).")
                    continue # Continue the loop without updating

                new_email = input(f"Enter new email address (current: {found_contact.email}, leave blank to keep current): ")

                # Call update_contact with the found contact's actual name to ensure it works with the function
                update_contact(contacts, found_contact.name, new_phone if new_phone else None, new_email if new_email else None)
            else:
                print(f"Contact '{name_to_update}' not found.")

        elif choice == '4':
            view_contacts(contacts)
        elif choice == '5': # New option for JSON export
            export_contacts_to_json(contacts)
        elif choice == '6':
            print("Exiting Contact Book. Goodbye!")
            break
        else:
            print("Invalid choice. Please enter a number between 1 and 6.")

# Run the main menu
main_menu()
```
