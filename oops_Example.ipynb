## Python OOPs Concepts

Object-Oriented Programming (OOP) is a programming paradigm based on the concept of "objects", which can contain data and code. The main principles of OOP are:

1.  **Class**: A blueprint or a template for creating objects. It defines a set of attributes and methods that the created objects will have.
2.  **Object**: An instance of a class. When a class is defined, no memory is allocated until an object of that class is created.
3.  **Encapsulation**: Bundling the data (attributes) and methods (functions) that operate on the data into a single unit or class. It also restricts direct access to some of an object's components, which can prevent accidental modification of data.
4.  **Inheritance**: A mechanism that allows a new class (subclass/child class) to inherit properties and behaviors from an existing class (superclass/parent class). This promotes code reusability.
5.  **Polymorphism**: The ability of an object to take on many forms. In Python, this often refers to method overriding (where a subclass provides a specific implementation of a method that is already defined in its superclass) and method overloading (though Python doesn't support method overloading in the traditional sense, it can be simulated).
6.  **Abstraction**: Hiding the complex implementation details and showing only the necessary and relevant information to the user. In Python, abstract classes and methods can be used to achieve this.

````python
class Dog:
    # Class attribute
    species = "Canis familiaris"

    # Constructor method
    def __init__(self, name, age):
        self.name = name  # Instance attribute
        self.age = age    # Instance attribute

    # Instance method
    def bark(self):
        return f"{self.name} says Woof!"

# Creating objects (instances of the Dog class)
my_dog = Dog("Buddy", 3)
your_dog = Dog("Lucy", 5)

print(f"My dog's name is {my_dog.name} and he is {my_dog.age} years old.")
print(f"My dog is a {my_dog.species}.")
print(my_dog.bark())

print(f"Your dog's name is {your_dog.name} and she is {your_dog.age} years old.")
print(f"Your dog is a {your_dog.species}.")
print(your_dog.bark())

# Inheritance example
class GoldenRetriever(Dog):
    def __init__(self, name, age, color):
        super().__init__(name, age) # Call parent class constructor
        self.color = color

    def bark(self): # Polymorphism: method overriding
        return f"{self.name} the {self.color} Golden Retriever says Arf!"

golden_dog = GoldenRetriever("Max", 2, "golden")
print(f"\n{golden_dog.name} is a {golden_dog.color} {golden_dog.species} and says {golden_dog.bark()}")
````

## Python Data Types

Python has several built-in data types, which are used to classify and manage different kinds of data. Here's a breakdown of the most common ones:

### 1. Numeric Types

*   **`int` (Integers)**: Whole numbers, positive or negative, without a decimal point. They can be of arbitrary precision.
*   **`float` (Floating-Point Numbers)**: Real numbers with a decimal point. They are represented using double-precision (64-bit) floating-point numbers.
*   **`complex` (Complex Numbers)**: Numbers with a real and an imaginary part, represented as `a + bj`, where `a` is the real part and `b` is the imaginary part.

### 2. Sequence Types

*   **`str` (Strings)**: A sequence of Unicode characters. Strings are immutable, meaning once created, their content cannot be changed.
*   **`list` (Lists)**: An ordered, mutable (changeable) sequence of items. Items can be of different data types and are enclosed in square brackets `[]`.
*   **`tuple` (Tuples)**: An ordered, immutable sequence of items. Similar to lists but cannot be changed after creation. Items are enclosed in parentheses `()`.
*   **`range` (Range)**: An immutable sequence of numbers, often used for looping a specific number of times.

### 3. Mapping Type

*   **`dict` (Dictionaries)**: An unordered, mutable collection of key-value pairs. Keys must be unique and immutable (like strings or numbers), while values can be of any data type. Enclosed in curly braces `{}`.

### 4. Set Types

*   **`set` (Sets)**: An unordered, mutable collection of unique items. Duplicate elements are not allowed. Enclosed in curly braces `{}`.
*   **`frozenset` (Frozen Sets)**: Similar to sets but are immutable. Once created, elements cannot be added or removed.

### 5. Boolean Type

*   **`bool` (Booleans)**: Represents one of two values: `True` or `False`. Used for logical operations.

### 6. None Type

*   **`NoneType` (`None`)**: A special data type that represents the absence of a value or a null value. It's often used as a placeholder.

````python
# Numeric Types
my_int = 10
my_float = 10.5
my_complex = 3 + 4j
print(f"Integer: {my_int}, Type: {type(my_int)}")
print(f"Float: {my_float}, Type: {type(my_float)}")
print(f"Complex: {my_complex}, Type: {type(my_complex)}")

# Sequence Types
my_string = "Hello Python"
my_list = [1, 2, "three", 4.0]
my_tuple = (10, 20, "thirty")
my_range = range(5) # 0, 1, 2, 3, 4
print(f"\nString: {my_string}, Type: {type(my_string)}")
print(f"List: {my_list}, Type: {type(my_list)}")
print(f"Tuple: {my_tuple}, Type: {type(my_tuple)}")
print(f"Range: {list(my_range)}, Type: {type(my_range)}")

# Mapping Type
my_dict = {"name": "Alice", "age": 30}
print(f"\nDictionary: {my_dict}, Type: {type(my_dict)}")

# Set Types
my_set = {1, 2, 3, 2, 1}
my_frozenset = frozenset([4, 5, 6])
print(f"\nSet: {my_set}, Type: {type(my_set)}")
print(f"Frozen Set: {my_frozenset}, Type: {type(my_frozenset)}")

# Boolean Type
my_bool_true = True
my_bool_false = False
print(f"\nBoolean True: {my_bool_true}, Type: {type(my_bool_true)}")
print(f"Boolean False: {my_bool_false}, Type: {type(my_bool_false)}")

# None Type
my_none = None
print(f"\nNone: {my_none}, Type: {type(my_none)}")
````

## Python Functions

A function is a block of organized, reusable code that performs a single, related action. Functions provide better modularity for your application and a high degree of code reusing.

### Defining a Function

In Python, you define a function using the `def` keyword, followed by the function name, parentheses `()`, and a colon `:`. The code block of the function is indented.

```python
def function_name(parameters):
    """Docstring (optional): explains what the function does."""
    # function body
    return expression # optional: returns a value
```

### Types of Arguments

1.  **Positional Arguments**: Arguments passed to a function in the correct positional order.
2.  **Keyword Arguments**: Arguments passed to a function by specifying the parameter name along with its value.
3.  **Default Arguments**: Arguments that have a default value in the function definition. If a value is not provided during the function call, the default value is used.
4.  **Arbitrary Arguments (`*args` and `**kwargs`)**: Allow a function to accept an arbitrary number of arguments.
    *   `*args`: Used to pass a non-keyworded, variable-length argument list.
    *   `**kwargs`: Used to pass a keyworded, variable-length argument list (as a dictionary).

### Lambda Functions (Anonymous Functions)

A lambda function is a small anonymous function. It can take any number of arguments, but can only have one expression. They are typically used for short, simple operations.

````python
# Basic Function
def greet(name):
    """This function greets the person passed in as a parameter."""
    return f"Hello, {name}!"

print(greet("Alice"))

# Function with Positional and Keyword Arguments
def describe_pet(animal_type, pet_name):
    return f"I have a {animal_type} named {pet_name}."

print(describe_pet("dog", "Buddy")) # Positional arguments
print(describe_pet(pet_name="Lucy", animal_type="cat")) # Keyword arguments

# Function with Default Arguments
def add(a, b=5):
    return a + b

print(add(10))     # b uses default value 5, output: 15
print(add(10, 20)) # b overrides default value, output: 30

# Function with Arbitrary Arguments (*args)
def sum_all(*numbers):
    total = 0
    for num in numbers:
        total += num
    return total

print(sum_all(1, 2, 3))         # Output: 6
print(sum_all(10, 20, 30, 40))  # Output: 100

# Function with Arbitrary Keyword Arguments (**kwargs)
def print_info(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

print_info(name="Bob", age=25, city="New York")

# Lambda Function
multiply = lambda x, y: x * y
print(f"\nLambda multiply(5, 3): {multiply(5, 3)}")

# Using lambda with higher-order functions (e.g., map, filter)
numbers = [1, 2, 3, 4, 5]
squared_numbers = list(map(lambda x: x**2, numbers))
print(f"Squared numbers (using map and lambda): {squared_numbers}")

even_numbers = list(filter(lambda x: x % 2 == 0, numbers))
print(f"Even numbers (using filter and lambda): {even_numbers}")
````
